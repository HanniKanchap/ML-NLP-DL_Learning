```python
from sklearn.datasets import load_iris
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Sequential
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical
```

```python
iris_data = load_iris()
```

```python
y = to_categorical(iris_data['target'],num_classes = 3)
```

```python
x_train,x_test,y_train,y_test = train_test_split(iris_data['data'],y,test_size=0.25,random_state=42)
```

```python
results = []
for activation_fn in ['relu','softmax','sigmoid','tanh']:
    model = Sequential()
    model.add(Dense(5,activation=activation_fn,input_shape = (4,)))
    model.add(Dense(4,activation=activation_fn))
    model.add(Dense(3,activation='softmax'))

    model.compile(optimizer = 'adam',loss = 'categorical_crossentropy',metrics = ['accuracy'])
    model.fit(x_train,y_train,epochs = 100)

    loss,accuracy = model.evaluate(x_test,y_test)
    results.append(f'Using {activation_fn}, the accuracy is {accuracy} and the loss is {loss}')

for el in results:
    print(el)
```
