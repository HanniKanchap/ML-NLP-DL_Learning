```python
import numpy as np
```

```python
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))

def sigmoid_derivative(z):
    return z*(1-z)

np.random.seed(42)
W1 = np.random.rand(2,2)
b1 = np.zeros((1,2))

W2 = np.random.rand(2,1)
b2 = np.zeros((1,1))
```

```python
lr = 0.1
epochs = 10000
```

```python
for epoch in range(epochs):
    z1 = np.dot(X,W1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1,W2) + b2
    a2 = sigmoid(z2)

    error = y - a2

    ## Backpropogation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update weights
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

# Final predictions
print("Predictions after training:")
print(a2.round(3))


```

```python
from sklearn.datasets import load_iris
from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Sequential
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical

iris_data = load_iris()
y = to_categorical(iris_data['target'],num_classes = 3)

x_train,x_test,y_train,y_test = train_test_split(iris_data['data'] ,y ,test_size=0.25 , random_state = 42 )

results = []

model = Sequential()
model.add(Dense(8,activation='relu',input_shape = (4,)))
model.add(Dense(10,activation='relu'))
model.add(Dense(3,activation='softmax'))

model.compile(optimizer = 'adam',loss = 'categorical_crossentropy',metrics = ['accuracy'])
model.fit(x_train,y_train,epochs = 50,batch_size=6)

loss,accuracy = model.evaluate(x_test,y_test)
print(f'Loss : {loss} , Accuracy : {accuracy}')

```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Input
from tensorflow.keras.utils import to_categorical
import numpy as np


data = load_breast_cancer()
X = data.data
y = data.target


y_onehot = to_categorical(y, num_classes=2)


X_train, X_test, y_train, y_test = train_test_split(X, y_onehot, test_size=0.2, random_state=42)

loss_functions = ['categorical_crossentropy', 'mean_squared_error', 'mean_absolute_error']

for loss_fn in loss_functions:
    print(f"\nTraining with loss function: {loss_fn}")

    model = Sequential()
    Input(shape = (30,))
    model.add(Dense(10, activation='relu'))
    model.add(Dense(8, activation='relu'))
    model.add(Dense(2, activation='softmax'))

    # Compile with chosen loss
    model.compile(optimizer='adam', loss=loss_fn, metrics=['accuracy'])

    # Train
    model.fit(X_train, y_train, epochs=30, batch_size=5, verbose=0)

    # Evaluate
    loss, acc = model.evaluate(X_test, y_test, verbose=0)
    print(f"Loss: {loss:.4f}, Accuracy: {acc:.3f}")
```
